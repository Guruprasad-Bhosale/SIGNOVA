# 03 — ISLTranslate Dataset Statistics & Distribution Analysis
Analyzes sentence length distributions, vocabulary frequency distributions, and split balances.

```python
import json
import pandas as pd
from pathlib import Path

# Load precomputed statistical summaries
with open('../outputs/reports/translation_statistics.json', 'r') as f:
    stats = json.load(f)

print(f"Total Valid Sentences: {stats['valid_sentence_count']}")
print(f"Unique Sentences: {stats['unique_sentence_count']}")
print(f"Vocabulary Size: {stats['vocabulary_size']}")
print(f"Mean Token Length: {stats['tokens_per_sentence']['mean']} words")
```

```python
# Inspect Split Balances
manifest = pd.read_csv('../data/manifests/isltranslate_manifest.csv')
print("Official 80/10/10 Split Breakdown:")
print(manifest['split'].value_counts(normalize=True) * 100)

# Inspect Proposed Session-Independent Split
prop_split = pd.read_csv('../data/manifests/proposed_signer_independent_split.csv')
print("\nProposed Session-Independent Split Breakdown:")
print(prop_split['proposed_independent_split'].value_counts(normalize=True) * 100)
```
